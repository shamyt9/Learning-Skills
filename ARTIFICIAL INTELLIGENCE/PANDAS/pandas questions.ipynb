### Select specific rows using iloc and loc.

### USING ILOC

```python
import pandas as pd 
s={
    "name":["Sham","Manish","Changesh"],
    "age":[12,23,34]
}
d=pd.DataFrame(s,index=['A','B','C'])
print(d.iloc[[0,1],[1]])
```

### using loc 

```python
import pandas as pd 
s={
    "name":["Sham","Manish","Changesh"],
    "age":[12,23,34]
}
d=pd.DataFrame(s,index=['A','B','C'])
print(d.loc[['A'],['age']])
```

### Retrieve rows where a string column contains a specific substring.

```python
s={
    "name":["Sham","Manish","Changesh"],
    "age":[12,23,34]
}
d=pd.DataFrame(s,index=['A','B','C'])
data=d[d["name"].str.contains('sh',case=False)]
print(data)
```

# Data Aggregation and Grouping

### Use groupby() to find the sum of a numerical column based on a categorical column.

```python
df=pd.read_csv('customers-100.csv')
sw=pd.DataFrame(df)
gd=sw.groupby('Country')['Index'].count()
print(gd)
```

### Find the mean, min, and max values for each group in a DataFrame.

```python
demo=pd.read_csv('demo.csv')
df=pd.DataFrame(demo)
gr=df.groupby('country')['salary'].min()
gr2=df.groupby('country')['salary'].max()
gr3=df.groupby('country')['salary'].mean()

print(gr)
print(gr2)
print(gr3)
```

## concating two rows

```python
import pandas as pd
data1={
    "name":["Sham","Sahil","Sadiya","Dhruv"],
    "course":["BCA","BCA","O LEVEL","PGDCA"]
}
data2={
    "name":["Subham","Manish","Shalini","Anusan"],
    "course":["BCA","BCA","O LEVEL","PGDCA"]
}
df1=pd.DataFrame(data1)
df2=pd.DataFrame(data2)

newDf=pd.concat([df1,df2],ignore_index=True)
print(newDf)

```

 ## merging the dataframe

```python
import pandas as pd

# Dataset 1: Customer Information
customer_df = pd.DataFrame({
    'Customer_ID': [1, 2, 3, 4, 5],
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Emily'],
    'City': ['New York', 'London', 'Paris', 'New York', 'London']
})

# Dataset 2: Order Information
order_df = pd.DataFrame({
    'Order_ID': [101, 102, 103, 104, 105],
    'Customer_ID': [1, 2, 1, 3, 5],
    'Order_Date': ['2023-01-15', '2023-02-20', '2023-03-10', '2023-04-05', '2023-05-12']
})

# Dataset 3: Product Information
product_df = pd.DataFrame({
    'Product_ID': [1, 2, 3],
    'Product_Name': ['Laptop', 'Phone', 'Tablet'],
    'Category': ['Electronics', 'Electronics', 'Electronics']
})
print(customer_df,end="\n\n\n\n")
print(order_df)
```

## PERFORM INNER JOIN

```python
mergedInner=pd.merge(customer_df,order_df,on="Customer_ID",how="inner")
print(mergedInner)
```

```python
mergedInner=pd.merge(customer_df,order_df,on="Customer_ID",how="outer")
print(mergedInner)
```
