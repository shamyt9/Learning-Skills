# FINDING MISSING VALUES

```python
import pandas as pd
l={
    "name":["alice",None,"Manish","Rohan","Manish","Aditya"],
    "Age":[12,None,34,23,32,15],
    "Course":["BCA",None,"PGDCA","BCA","MCA","BCA"]
}
ps=pd.DataFrame(l)
print(ps.isnull())
print("Total missing values: ",ps.isnull().sum())
```

# REMOVING ANY MISSING VALUE in ROWS

```python
ps.dropna(axis=0,inplace=True)
ps["none"]=[None,None,None,None,None]
print(ps)
```

# REMOVING MISSING VALUES IN COLUMNS

```python
ps.dropna(axis=1,inplace=True)
print(ps)
```

# REMOVING ALL THE MISSING VALUES

```python
import pandas as pd
l={
    "name":["alice",None,"Manish","Rohan",None,"Aditya"],
    "Age":[12,None,34,23,32,15],
    "Course":["BCA",None,"PGDCA","BCA","MCA","BCA"]
}
pr=pd.DataFrame(l)
print(pr)
print("Total missing values: ",pr.isnull().sum())
```

```python
pr.dropna(inplace=True)
print(pr)
```

# FILLING VALUES IN PLACE OF MISSING VALUES

```python
import pandas as pd
l={
    "name":["alice",None,"Manish","Rohan",None,"Aditya"],
    "Age":[12,None,34,23,32,15],
    "Course":["BCA",None,"PGDCA","BCA","MCA","BCA"]
}
pp=pd.DataFrame(l)
print(pp)
```

# USING INTERPOLATION 

```python
new={
    "name":["name1","name2","name3","name4","name5"],
    "AGE":[12,23,None,45,56],
    "SALARY":[10000,None,23000,34000,1234]
}
pdf=pd.DataFrame(new)

pdf["AGE"]=pdf["AGE"].interpolate(method="linear")

print(pdf)
```

# sorting and aggregation

```python
employee = {
    "name": ["Manish Goswami","Alice Wonderland", "Bob the Builder", "Charlie Brown", "David Copperfield", "Emily Dickinson"],
    "id": ["EMP1000","EMP1001", "EMP1002", "EMP1003", "EMP1004", "EMP1005"],
    "salary": [23000,75000, 62000, 80000, 58000, 70000],
    "department": ["FINANCE","Marketing", "IT", "Finance", "HR", "Sales"]
}
pd2=pd.DataFrame(employee)
pd2.sort_values(by="department",inplace=True)
print(pd2)
```

# MERGING AND JOINING

```python
import pandas as pd

# Dataset 1: Customer Information
customer_df = pd.DataFrame({
    'Customer_ID': [1, 2, 3, 4, 5],
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Emily'],
    'City': ['New York', 'London', 'Paris', 'New York', 'London']
})

# Dataset 2: Order Information
order_df = pd.DataFrame({
    'Order_ID': [101, 102, 103, 104, 105],
    'Customer_ID': [1, 2, 1, 3, 5],
    'Order_Date': ['2023-01-15', '2023-02-20', '2023-03-10', '2023-04-05', '2023-05-12']
})

# Dataset 3: Product Information
product_df = pd.DataFrame({
    'Product_ID': [1, 2, 3],
    'Product_Name': ['Laptop', 'Phone', 'Tablet'],
    'Category': ['Electronics', 'Electronics', 'Electronics']
})
print(customer_df,end="\n\n\n\n")
print(order_df)
```

## INNSER JOIN

```python
mergedInner=pd.merge(customer_df,order_df, on="Customer_ID", how="inner")
print(mergedInner)
```

### OUTER

```python
mergedInner=pd.merge(customer_df,order_df, on="Customer_ID", how="outer")
print(mergedInner)
```

### LEFT

```python
mergedInner=pd.merge(customer_df,order_df, on="Customer_ID", how="left")
print(mergedInner)
```

### RIGHT

```python
mergedInner=pd.merge(customer_df,order_df, on="Customer_ID", how="right")
print(mergedInner)
```

# MERGING DATA SETS

```python
import pandas as pd

# Dataset 1: Students
students_df = pd.DataFrame({
    'Student_ID': [1, 2, 3],
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Grade': ['A', 'B', 'A']
})

# Dataset 2: Courses
courses_df = pd.DataFrame({
    'Course_ID': [101, 102, 103],
    'Course_Name': ['Math', 'Science', 'English']
})

# Dataset 3: Student_Courses (Linking Students and Courses)
student_courses_df = pd.DataFrame({
    'Student_ID': [1, 1, 2, 3],
    'Course_ID': [101, 102, 101, 103]
})

print(students_df,end="\n\n\n\n\n")
print(student_courses_df,end="\n\n\n\n\n")
print(courses_df,end="\n\n\n\n\n")
```

```python
concate=pd.concat([students_df,student_courses_df,courses_df],axis=1,ignore_index=True)
print(concate)
```
