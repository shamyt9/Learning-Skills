```python
data2 = {
    "name": [
        "Ram", "Shyam", "Ghanshyam", "Rita", "Anita", "Vikas", "Raj", "Priya", "Sunil", "Amit", 
        "Neha", "Anil", "Kavita", "Suresh", "Ramesh", "Meera", "Kiran", "Arun", "Sneha", "Pooja"
    ],
    "age": [
        25, 23, 34, 28, 30, 24, 26, 27, 35, 31, 
        29, 33, 22, 32, 26, 30, 27, 36, 24, 28
    ],
    "department": [
        "HR", "Finance", "IT", "Marketing", "Operations", "IT", "Finance", "HR", "Marketing", "Operations",
        "HR", "Finance", "IT", "Marketing", "Operations", "IT", "Finance", "HR", "Marketing", "Operations"
    ],
    "salary": [
        50000, 55000, 60000, 52000, 58000, 51000, 57000, 53000, 62000, 59000, 
        54000, 61000, 56000, 63000, 58000, 52000, 57000, 64000, 55000, 59000
    ],
    "experience_years": [
        2, 1, 5, 3, 4, 2, 3, 2, 6, 4, 
        3, 5, 2, 6, 4, 3, 3, 7, 2, 4
    ]
}

df=pd.DataFrame(data2)
print(df)
```

# FILTERING ROWS ON SINGLE CONDITION

```python
highPayingSalary=df[df["salary"]>50000]
print(f"HIGH PAYING SALARIES AR: {highPayingSalary}")
```

# FILTERING DATA ON MULTIPLE CONDITION

```python
filt=df[(df["salary"]>50000) & (df["experience_years"]<4)]
print("USING AND CONDITION")
print(filt)


print("USING OR CONDITION")
filt=df[(df["salary"]>50000) | (df["experience_years"]<4)]
print(filt)
```
