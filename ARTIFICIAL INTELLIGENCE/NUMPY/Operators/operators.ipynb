## Use of "nditer" function to escape nested loop for multiple dimension arrays

```python
import numpy as np
v=np.array([[1,2,3,4],[3,4,5,50]])

for i in np.nditer(v):
    print(i)
```

## ndenumerate() function use to iterate multi-dimen array with indesing

```python
import numpy as np
v=np.array([[1,2,3,4],[3,4,5,50]])

for i,d in np.ndenumerate(v):
    print("index",i,": ",d)
```

# COPY VS VIEW

### changing in view and copy 

```python
import numpy as np
v1=np.array([[1,2,3,4],[3,4,5,50]])
print(v1)
print()
v2=v1.copy()
v2[0][0]=50
print(v1)
print(v2)


v3=np.array([[1,2,3,4],[3,4,5,50]])
print(v3)
print()
v4=v3.view()
v4[0][0]=90
print(v4)
print(v3)
```

## MATRIX FUNCTIONS

## transpose()

```python
import numpy as np
v=np.matrix([[1,3,2],[23,12,45]])

print("BEFORE TRANSPOSE:  ")
print(v)
print("**********************************************")
print("AFTER TRANSPOSE:   ")
print(np.transpose(v)) # can be written as v.T
```

## inverse()

```python
import numpy as np
v=np.matrix([[1,3],[23,12]])

print("BEFORE INVERSE:  ")
print(v)
print("**********************************************")
print("AFTER INVERSE:   ")
print(np.linalg.inv(v)) 
```

## power

```python
import numpy as np
v=np.matrix([[1,3],[23,12]])

print("BEFORE POWER:  ")
print(v)
print("**********************************************")
print("AFTER POWER:   ")
print(np.linalg.matrix_power(v,3)) 
```

## determinant of matrix

```python
import numpy as np
v=np.matrix([[1,1],[23,12]])

print("BEFORE POWER:  ")
print(v)
print("**********************************************")
print("AFTER POWER:   ")
print(np.linalg.det(v)) 
```
