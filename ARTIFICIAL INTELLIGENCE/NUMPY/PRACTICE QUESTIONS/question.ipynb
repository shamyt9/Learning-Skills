## Q2-: Write a function to count the number of vowels in a string.				


```python
def countVowels(str1):
    str1=str1.lower()
    
    count=0
    l1=['a','e','i','o','u']
    for x in str1:
        if x in l1:
            count+=1
            
    print("Number of vowels : ",count)   

str1=input("Enter your Statement: ")
countVowels(str1)
```

## Q3-: Write a function to check if two strings are anagrams.


```python
def anagrams(str1,str2):
    print("----------------------------------------------------")
 
    str1=str1.lower()
    str2=str2.lower()
    print("Word 1: "+str1)
    print("Word 2: "+str2)
    print("----------------------------------------------------")
    count1=len(str1)
    count2=len(str2)
    if(count1==count2):
        st1=list(str1)
        st2=list(str2)
        st1.sort()
        st2.sort()
        if(st1==st2):
            st1="".join(st1)
            st2="".join(st2)
            print(st1," and ",st2," are anagrams.")
        else:
            print("Not")

    else:
        print("No")
            
    
    
    
    


str1=input("Enter String 1: ")    
str2=input("Enter String 2: ")   
anagrams(str1,str2)
```

## Q7-: Write a function to merge two dictionaries, adding values for common keys.


```python
def mergeDict(dict1,dict2):
    newdict={}
    for key,value in dict1.items():
        newdict[key]=value

    for key,value in dict2.items():
        if key in newdict:
            newdict[key]+=value
        else:
            newdict[key]=value

    print(newdict)

dict1={'a':12,'b':123,'c':45}
dict2={'a':12,'r':23,'y':45}
mergeDict(dict1,dict2)
```

## Q13-:Write a function that returns the factorial of a number using recursion.


```python
def returnFact(n):
    if n==0 or n==1:
        return 1
    else:
        return n*returnFact(n-1)
getFact=returnFact(4)
print(getFact)
        
```

## Q17-:Write a generator function that yields the first n Fibonacci numbers.0,1,1,2,3,5,8,13


```python
def fibn(n):
    track=1
    i=0
    x=1
    yield i
    while n>=track:
        yield x
        i,x=x,x+i
        track+=1
get=fibn(9)
print(next(get))
print(next(get))
print(next(get))
print(next(get))
print(next(get))
print(next(get))
print(next(get))
print(next(get))
print(next(get))
        
        
```

## Q18-: Write a recursive function to generate permutations of a string.


```python
import itertools

def perms(s):
    if len(s)==1:
        return [s]
    else:
        return [''.join(p) for p in itertools.permutations(s)]

x=perms("word")
print(x)
        
```
