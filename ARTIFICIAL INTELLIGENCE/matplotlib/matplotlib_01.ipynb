# 02_Using dataframe

```python
import pandas as pd

data={
    "name":["Shrikant","Jaikant","Mahakant"],
    "Salary":[20000,30000,40000],
    "Age":[23,34,45]
}
df=pd.DataFrame(data)
print(df)
plt2=mtp.plot(df["name"],df["Salary"])
mtp.title("Line Chart")
mtp.legend("Line")
plt2
```

## 01.02_ BAR PLOTTING

```python
import pandas as pd

data={
    "name":["Shrikant","Jaikant","Mahakant"],
    "Salary":[20000,30000,40000],
    "Age":[23,34,45]
}
df=pd.DataFrame(data)
print(df)
plt2=mtp.bar(df["name"],df["Salary"])
mtp.title("Bar Chart")
mtp.legend("BARS")
plt2
```
