```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as mtp
from sklearn.model_selection import train_test_split
dataSet=pd.read_csv("Salary_Data.csv")
df=pd.DataFrame(dataSet)
X=df.iloc[:,:-1].values
Y=df.iloc[:,-1].values
X_train,X_test, Y_train, Y_test= train_test_split(X,Y,test_size=1/5,random_state=0)
print(X_test)
print("-------------")
print(X_train)
print("-------------")
print(Y_test)
print("-------------")
print(Y_train)
#pt=mtp.plot(X,Y)
```

```python
from sklearn.linear_model import LinearRegression
regressor=LinearRegression()
regressor.fit(X_train,Y_train)
mtp.scatter(X_train,Y_train,color='blue')
mtp.plot(X_train,regressor.predict(X_train),color='red')
Y_pred=regressor.predict(X_test)
```

```python
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
mse= mean_squared_error(Y_test,Y_pred)
rsm=np.sqrt(mse)
print(rsm)
#TELLS THE SCORE OF MODEL CORRECTNESS
r2 = r2_score(Y_test, Y_pred)
print("R-squared:", r2)
```

```python
mtp.scatter(Y_test, Y_pred, color='blue')
mtp.plot([Y_test.min(), Y_test.max()], [Y_test.min(), Y_test.max()], color='red', linestyle='--')
mtp.xlabel("Actual Values")
mtp.ylabel("Predicted Values")
mtp.title("Actual vs Predicted")
mtp.show()
```
