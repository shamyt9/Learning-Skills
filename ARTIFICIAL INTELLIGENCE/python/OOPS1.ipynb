```python
class Car:
    def __init__(self,brand,model):
     self.__brand=brand
     self.model=model
    def letsView(self):
        return f"{self.brand} {self.model}"
       

##ELECTRIC CAR CLASS
class ElectricCar(Car):
    def __init__(self,brand,model,bs):
        super().__init__(brand,model)
        self.bs=bs
        

myCar=ElectricCar("Toyota","Fortuner","2000KHAZ")
print(myCar.model,myCar.bs)
```
